```python
from sklearn.metrics import precision_recall_curve

def recall_at_precision(y_true, y_score, min_precision=0.90):
    p, r, _ = precision_recall_curve(y_true, y_score)
    ok = p[:-1] >= min_precision
    return r[:-1][ok].max() if ok.any() else 0.0
```

```python
# one test set, split off first, stratified (or grouped / time-ordered if R1 found groups or time)


from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.model_selection import StratifiedKFold
CV = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
from sklearn.metrics import make_scorer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import loguniform, randint

scorer = make_scorer(recall_at_precision, response_method="predict_proba")
N_ITER = 20

models = {
    "logreg": (
        LogisticRegression(max_iter=5000, random_state=42),
        {"clf__C": loguniform(1e-3, 1e2)},
    ),
    "rf": (
        RandomForestClassifier(random_state=42),
        {"clf__n_estimators": randint(100, 500),
         "clf__max_depth": randint(2, 12),
         "clf__min_samples_leaf": randint(1, 10)},
    ),
}

results = {}
for name, (clf, dist) in models.items():
    pipe = Pipeline([("scale", StandardScaler()), ("clf", clf)])  # same preprocessing for all
    search = RandomizedSearchCV(pipe, dist, n_iter=N_ITER, scoring=scorer,
                                cv=CV, random_state=42, n_jobs=-1)
    search.fit(X_train, y_train)
    results[name] = search
    print(f"{name}: CV recall@P>=0.90 = {search.best_score_:.3f}  {search.best_params_}")
```

```python
for name, s in results.items():
    i = s.best_index_
    r = s.cv_results_
    print(f"{name}: mean={r['mean_test_score'][i]:.3f}  std={r['std_test_score'][i]:.3f}")
```

    var_smoothing. np.allclose against GaussianNB's predict_proba.
[ ] The tuned SVM's decision_function reproduced from support_vectors_,
    dual_coef_ and intercept_, for its kernel. np.allclose against
    SVC.decision_function.

```python
SEED = 42;

from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

scorer = make_scorer(recall_at_precision,
                     response_method=("predict_proba", "decision_function"))

models = {
    "logreg": (LogisticRegression(max_iter=5000, random_state=SEED),
               {"clf__C": loguniform(1e-3, 1e2)}),
    "rf": (RandomForestClassifier(random_state=SEED),
           {"clf__n_estimators": randint(100, 500),
            "clf__max_depth": randint(2, 12),
            "clf__min_samples_leaf": randint(1, 10)}),
    "knn": (KNeighborsClassifier(),
            {"clf__n_neighbors": randint(1, 31),
             "clf__weights": ["uniform", "distance"],
             "clf__p": [1, 2]}),
    "gnb": (GaussianNB(),
            {"clf__var_smoothing": loguniform(1e-12, 1e-1)}),
    "svm": (SVC(random_state=SEED),
            {"clf__C": loguniform(1e-2, 1e3),
             "clf__kernel": ["rbf", "linear"],
             "clf__gamma": loguniform(1e-4, 1e0)}),
}

results = {}
for name, (clf, dist) in models.items():
    pipe = Pipeline([("scale", StandardScaler()), ("clf", clf)])
    s = RandomizedSearchCV(pipe, dist, n_iter=N_ITER, scoring=scorer,
                           cv=CV, random_state=SEED, n_jobs=-1)
    s.fit(X_train, y_train)
    results[name] = s
    i = s.best_index_
    print(f"{name}: {s.cv_results_['mean_test_score'][i]:.3f} ± "
          f"{s.cv_results_['std_test_score'][i]:.3f}  {s.best_params_}")
```

```python
import numpy as np
from scipy.special import logsumexp

sc = results["knn"].best_estimator_.named_steps["scale"]
Xtr, Xte = sc.transform(X_train), sc.transform(X_test)   # scaled, as in the pipeline
ytr = y_train.to_numpy()
```

```python
#KNN, vectorised

def knn_predict(Xtr, ytr, Xte, k, p, weights):
    D = (np.abs(Xte[:, None, :] - Xtr[None, :, :]) ** p).sum(-1) ** (1 / p)
    idx = np.argpartition(D, k - 1, axis=1)[:, :k]
    d = np.take_along_axis(D, idx, 1)
    if weights == "uniform":
        w = np.ones_like(d)
    else:                                    # sklearn: exact matches dominate
        z = d == 0
        w = np.where(z.any(1, keepdims=True), z.astype(float), 1 / np.where(z, 1, d))
    votes = (w[..., None] * np.eye(2)[ytr[idx]]).sum(1)
    return votes.argmax(1)

knn = results["knn"].best_estimator_.named_steps["clf"]
mine = knn_predict(Xtr, ytr, Xte, knn.n_neighbors, knn.p, knn.weights)
print(knn.n_neighbors, knn.p, knn.weights, "| match:", np.array_equal(mine, knn.predict(Xte)))
```

```python
def gnb_fit(X, y, var_smoothing):
    classes = np.unique(y)
    eps = var_smoothing * X.var(axis=0).max()
    theta = np.array([X[y == c].mean(0) for c in classes])
    var = np.array([X[y == c].var(0) for c in classes]) + eps
    prior = np.array([(y == c).mean() for c in classes])
    return theta, var, prior

def gnb_predict_proba(X, theta, var, prior):
    jll = (np.log(prior)
           - 0.5 * np.log(2 * np.pi * var).sum(1)
           - 0.5 * (((X[:, None, :] - theta) ** 2) / var).sum(2))
    return np.exp(jll - logsumexp(jll, axis=1, keepdims=True))

gnb = results["gnb"].best_estimator_.named_steps["clf"]
theta, var, prior = gnb_fit(Xtr, ytr, gnb.var_smoothing)
print("allclose:", np.allclose(gnb_predict_proba(Xte, theta, var, prior), gnb.predict_proba(Xte)))
```

```python
def svm_decision(X, svm):
    SV = svm.support_vectors_
    if svm.kernel == "linear":
        K = X @ SV.T
    elif svm.kernel == "rbf":
        sq = (X ** 2).sum(1)[:, None] + (SV ** 2).sum(1)[None, :] - 2 * X @ SV.T
        K = np.exp(-svm.gamma * sq)
    else:
        raise ValueError(svm.kernel)
    return K @ svm.dual_coef_.ravel() + svm.intercept_[0]

svm = results["svm"].best_estimator_.named_steps["clf"]
print(svm.kernel, "| allclose:", np.allclose(svm_decision(Xte, svm), svm.decision_function(Xte)))
```

```python
#R4

import pandas as pd

rows = []
for name, s in results.items():
    r = s.cv_results_
    i = s.best_index_
    best, sd = r["mean_test_score"][i], r["std_test_score"][i]
    within = int((r["mean_test_score"] >= best - sd).sum())   # includes the best itself
    rows.append({"model": name, "best_cv": round(best, 3), "std": round(sd, 3),
                 "within_1sd": f"{within}/{N_ITER}"})

summary = pd.DataFrame(rows).set_index("model")
summary
```

All models use RandomizedSearchCV, N_ITER = 20, shared CV object,
shared pipeline. Continuous parameters (C, gamma, var_smoothing) are log-uniform.
KNN metric is tuned via p (Manhattan/Euclidean). SVM kernels: rbf, linear
(gamma is ignored for linear, so fewer distinct linear candidates).
Within one std of the best: logreg 18/20, rf 19/20, knn 12/20, gnb 20/20,
svm 5/20. Top models are within one std of each other; best scores are
optimistic (max of 20 candidates on the same folds).

```python
#R5
from sklearn.base import clone
from sklearn.model_selection import RepeatedStratifiedKFold, cross_val_score

RCV = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=SEED)  # one object, 25 folds, every model

scores = {
    name: cross_val_score(clone(s.best_estimator_), X_train, y_train,
                          scoring=scorer, cv=RCV, n_jobs=-1)
    for name, s in results.items()
}
pd.DataFrame({n: [v.mean().round(3), v.std(ddof=1).round(3)] for n, v in scores.items()},
             index=["mean", "std"])
```

```python
from itertools import combinations

# Rule, fixed before looking: "clearly different" only if |mean diff| > std of the
# difference AND the leader is strictly ahead on at least 20 of 25 folds.
rows = []
for a, b in combinations(scores, 2):
    d = scores[a] - scores[b]
    m, sd = d.mean(), d.std(ddof=1)
    wa, wb, ties = int((d > 0).sum()), int((d < 0).sum()), int((d == 0).sum())
    lead, other = (a, b) if m > 0 else (b, a)
    if abs(m) > sd and max(wa, wb) >= 20:
        verdict = f"{lead} clearly better than {other}"
    else:
        verdict = f"{a} vs {b}: not shown to be different"
    rows.append({"pair": f"{a} - {b}", "mean_diff": round(m, 3), "std_diff": round(sd, 3),
                 f"folds_a_ahead": wa, "folds_b_ahead": wb, "ties": ties, "verdict": verdict})

pd.DataFrame(rows).set_index("pair")
```

```python
#r6

from sklearn.metrics import roc_auc_score, average_precision_score, confusion_matrix

def get_score(est, X):
    return est.predict_proba(X)[:, 1] if hasattr(est, "predict_proba") else est.decision_function(X)

rows, cms = [], {}
for name, s in results.items():
    est = s.best_estimator_                      # already refit on all of X_train
    sc = get_score(est, X_test)
    tn, fp, fn, tp = confusion_matrix(y_test, est.predict(X_test)).ravel()   # default threshold
    cms[name] = [[tn, fp], [fn, tp]]
    i = s.best_index_
    rows.append({
        "model": name,
        "recall@P>=0.90": round(recall_at_precision(y_test, sc), 3),
        "cv_recall@P>=0.90": round(s.cv_results_["mean_test_score"][i], 3),
        "roc_auc": round(roc_auc_score(y_test, sc), 3),
        "pr_auc": round(average_precision_score(y_test, sc), 3),
        "missed (FN)": int(fn), "false alarms (FP)": int(fp),
        "of positives": int(fn + tp),
    })

test_table = pd.DataFrame(rows).set_index("model")
test_table
```

```python
#R7
import numpy as np
from scipy.special import logsumexp   # not needed here, harmless
from sklearn.base import clone

fit = {n: s.best_estimator_ for n, s in results.items()}
clf = {n: e.named_steps["clf"] for n, e in fit.items()}
nf, ntr = X_train.shape[1], len(X_train)
Xte_s = fit["knn"].named_steps["scale"].transform(X_test)   # same scaler in every pipeline
ytr, yt = y_train.to_numpy(), y_test.to_numpy()

stored = {
    "knn":    clf["knn"].n_samples_fit_ * nf + clf["knn"].n_samples_fit_,
    "gnb":    clf["gnb"].theta_.size + clf["gnb"].var_.size + clf["gnb"].class_prior_.size,
    "svm":    clf["svm"].support_vectors_.size + clf["svm"].dual_coef_.size + clf["svm"].intercept_.size,
    "logreg": clf["logreg"].coef_.size + clf["logreg"].intercept_.size,
    "rf":     sum(t.tree_.node_count for t in clf["rf"].estimators_),
}
n_sv = clf["svm"].support_vectors_.shape[0]
work = {
    "knn":    f"{ntr} distances ({nf} dims each)",
    "gnb":    f"{2 * nf} log-densities",
    "svm":    f"{n_sv} kernel evaluations ({nf} dims each)",
    "logreg": f"{nf} multiply-adds",
    "rf":     f"{clf['rf'].decision_path(Xte_s)[0].sum() / len(X_test):.0f} node tests on average",
}
pd.DataFrame({"numbers_stored": stored, "work_per_prediction": work})
```

```python
knn_pred = fit["knn"].predict(X_test)
cand = np.where((yt == 1) & (knn_pred == 0))[0]
j = int(cand[0]) if len(cand) else int(np.where(yt == 1)[0][0])
x = Xte_s[j]
print("test row", j, "| true label:", yt[j])

# KNN: nearest training cases and labels
dist, idx = clf["knn"].kneighbors(Xte_s[[j]])
display(pd.DataFrame({"train_row": X_train.index[idx[0]], "distance": dist[0].round(3),
                      "label": ytr[idx[0]]}))

# GNB: per-feature log-likelihood contributions
g = clf["gnb"]
ll = -0.5 * np.log(2 * np.pi * g.var_) - 0.5 * (x - g.theta_) ** 2 / g.var_
gc = pd.DataFrame({"benign_ll": ll[0], "malignant_ll": ll[1],
                   "log_odds_malignant": ll[1] - ll[0]}, index=X.columns)
print("prior log-odds:", np.log(g.class_prior_[1] / g.class_prior_[0]).round(2),
      "| sum of feature log-odds:", gc.log_odds_malignant.sum().round(2))
display(gc.reindex(gc.log_odds_malignant.abs().sort_values(ascending=False).index).head(8).round(2))

# SVM: what it can say about this case
s = clf["svm"]
print(s.kernel, "| decision value:", s.decision_function(Xte_s[[j]])[0].round(3))
if s.kernel == "linear":
    w = (s.dual_coef_ @ s.support_vectors_).ravel()
    display(pd.Series(w * x, index=X.columns).sort_values(key=abs, ascending=False).head(8).round(2))
else:
    k = np.exp(-s.gamma * ((s.support_vectors_ - x) ** 2).sum(1))
    c = s.dual_coef_.ravel() * k
    top = np.argsort(-np.abs(c))[:8]
    display(pd.DataFrame({"train_row": X_train.index[s.support_[top]],
                          "label": ytr[s.support_[top]], "contribution": c[top].round(3)}))
```

```python
# Cell 3: calibration
from sklearn.metrics import log_loss

svm_p = clone(fit["svm"]).set_params(clf__probability=True).fit(X_train, y_train)
probs = {"knn": fit["knn"].predict_proba(X_test)[:, 1],
         "gnb": fit["gnb"].predict_proba(X_test)[:, 1],
         "svm (Platt)": svm_p.predict_proba(X_test)[:, 1]}

rows = []
for n, p in probs.items():
    conf = (p > 0.99) | (p < 0.01)
    wrong = conf & ((p >= 0.5).astype(int) != yt)
    rows.append({"model": n, "log_loss": round(log_loss(yt, p), 3),
                 "confident_share": round(conf.mean(), 3), "n_confident": int(conf.sum()),
                 "confident_and_wrong": int(wrong.sum())})
pd.DataFrame(rows).set_index("model")
```

*R8. The decision**
WE CHOOSE      Logistic regression (scaled, C = 0.0746 as tuned).

BECAUSE        Evidence (5x5 repeated folds on the training data):
               logreg 0.978 ± 0.020 vs SVM 0.987 ± 0.017 on recall@P>=0.90.
               Difference -0.009 (std 0.016); the SVM was ahead on 9 folds,
               logreg on 1, with 15 ties. Not shown to be different, though
               the SVM leans ahead. Logreg vs KNN: +0.008 (std 0.023), 5 vs 1
               folds, 19 ties: not shown to be different. No pair of models
               met the rule (|diff| > std and >= 20 of 25 folds). These folds
               reuse the tuning data and are not independent, so even this
               is optimistic about certainty.
               Test set (42 malignant cases, scored once): logreg tied-best
               on recall@P>=0.90 (0.976), best ROC-AUC (0.998), missed 2 of 42
               at the default threshold with 0 false alarms (SVM 4, KNN 6).
               One case is 0.024 of recall, so the top models cannot be
               ranked reliably on this test set.
               Properties (R7): every prediction needs an explanation. Logreg
               stores 31 numbers, needs 30 multiply-adds per prediction, and
               gives a per-feature breakdown (coefficient x scaled value).
               The tuned SVM is rbf, so it can only point to support vectors;
               KNN stores 14,105 numbers and needs 455 distances; GNB's
               breakdown double-counts correlated features and it is the most
               overconfident model (log loss 0.500, 5 confident errors).
               Logreg calibration: log loss __, confident and wrong __.

WE GIVE UP     The SVM's slightly higher recall in CV (0.987 vs 0.978) and
               test-time calibration after Platt scaling (log loss 0.095).
               KNN's explanation by actual similar cases. Nothing measured
               from RF (lower on every fold comparison) or GNB.

WE WOULD
RECONSIDER IF  1. A larger evaluation set (at least 150 malignant cases) shows the
               SVM ahead of logreg on recall@P>=0.90 by 0.03 or more over
               repeated folds at the same precision floor.
               2. Logreg's test log loss exceeds the SVM's by more than 0.05,
               or it makes any confident error (>0.99 or <0.01).
               3. Reviewers prefer case-based explanations (KNN) to
               per-feature ones.
               4. The precision floor changes from 0.90 (rerun the metric).

```python
#R9

from sklearn.model_selection import StratifiedKFold, cross_val_score

OUTER = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)   # new outer splitter; inner is the shared CV

nested = {}
for name, (clf_, dist) in models.items():
    pipe = Pipeline([("scale", StandardScaler()), ("clf", clf_)])
    inner = RandomizedSearchCV(pipe, dist, n_iter=N_ITER, scoring=scorer,
                               cv=CV, random_state=SEED, n_jobs=-1)
    nested[name] = cross_val_score(inner, X_train, y_train, scoring=scorer, cv=OUTER)

rows = []
for name, s in results.items():
    i = s.best_index_
    r4 = s.cv_results_["mean_test_score"][i]
    nn = nested[name]
    rows.append({"model": name, "R4_best_cv": round(r4, 3),
                 "R5_repeated": round(scores[name].mean(), 3),
                 "nested_mean": round(nn.mean(), 3), "nested_std": round(nn.std(ddof=1), 3),
                 "shift_vs_R4": round(nn.mean() - r4, 3)})
pd.DataFrame(rows).set_index("model")
```
